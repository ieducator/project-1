```python
from transformers import pipeline

# Läs in klassificeringsmodellen
classifier = pipeline("zero-shot-classification")

print("✓ Modellen är redo!")
```

```python
# Definiera klassificeringskategorierna
kategorier = ["Rimlig AI-möjlighet", "Överdrivet anspråk/Hype"]

# Testa med ett enkelt påstående
pastand = "AI kan diagnosticera cancer från röntgenbilder"

# Klassificera
resultat = classifier(pastand, kategorier)

# Visa resultatet
print(f"Påstående: {pastand}")
print(f"Klassificering: {resultat['labels'][0]}")
print(f"Säkerhet: {round(resultat['scores'][0] * 100, 1)}%")
```

```python
# Lista med AI-påstanden från seminariet
pastand_lista = [
    "AI är helt objektiv och saknar bias",
    "AI kan diagnosticera cancer från röntgenbilder",
    "AI löser alla problem om vi bara ger den mer data",
    "Superintelligens kommer om 5 år",
    "AI kan förbättra sjukvård genom att identifiera mönster",
    "Machine Learning är bara statistik",
    "AI kan förutsäga väder en månad i förväg",
    "Neural Networks härmar hur hjärnan fungerar",
    "AI kan simulera mänskligt samtal",
    "Vissa AI-system kan tränas på data som innehåller bias"
]

kategorier = ["Rimlig AI-möjlighet", "Överdrivet anspråk/Hype"]

print("=" * 70)
print("KLASSIFICERING AV AI-PÅSTANDEN")
print("=" * 70)

for i, pastand in enumerate(pastand_lista, 1):
    resultat = classifier(pastand, kategorier)
    klassificering = resultat['labels'][0]
    sakerhet = round(resultat['scores'][0] * 100, 1)

    print(f"\n{i}. {pastand}")
    print(f"   Klassificering: {klassificering}")
    print(f"   Säkerhet: {sakerhet}%")
```

```python
kategorier = ["Rimlig AI-möjlighet", "Överdrivet anspråk/Hype"]

print("=" * 70)
print("KLASSIFICERA DITT AI-PÅSTÅENDE")
print("=" * 70)

# Mata in påståndet
pastand = input("\nSkriv ditt AI-påstående: ")

# Klassificera
resultat = classifier(pastand, kategorier)

klassificering = resultat['labels'][0]
sakerhet = round(resultat['scores'][0] * 100, 1)
motsatt_sakerhet = round((1 - resultat['scores'][0]) * 100, 1)

# Visa resultatet på ett pedagogiskt sätt
print("\n" + "=" * 70)
print(f"DIN KLASSIFICERING:")
print("=" * 70)
print(f"Påstående: {pastand}")
print(f"\nSystemets klassificering: {klassificering}")
print(f"Säkerhet: {sakerhet}%")
print(f"(Motsatt klassificering: {motsatt_sakerhet}%)")
print("=" * 70)

# Pedagogisk följdfråga
print("\nReflektionsfråga:")
print("• Var klassificeringen överraskande för dig?")
print("• Vilka antaganden ligger bakom påståendet?")
print("• Är klassificeringen helt tydlig eller beror det på kontext?")
```
